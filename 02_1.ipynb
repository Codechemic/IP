```python
from google.colab.patches import cv2_imshow
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
```

```python
origin_img = cv2.imread('/content/drive/MyDrive/Colab/images/lena256.jpg')
cv2_imshow(origin_img)
```

```python
RGB_img = cv2.cvtColor(origin_img, cv2.COLOR_BGR2RGB)
cv2_imshow(RGB_img)
```

```python
Red_img, Green_img, Blue_img = cv2.split(RGB_img)
cv2_imshow(Red_img)
cv2_imshow(Green_img)
cv2_imshow(Blue_img)
```

```python
plt.title('Original Image')
plt.imshow(RGB_img)
plt.show()
```

```python
RGB_img[:, :, 0] = Red_img
RGB_img[:, :, 1] = Red_img
RGB_img[:, :, 2] = Red_img
plt.subplot(1, 4, 1)
plt.title('Red Img')
plt.imshow(RGB_img)
RGB_img[:, :, 0] = Green_img
RGB_img[:, :, 1] = Green_img
RGB_img[:, :, 2] = Green_img
plt.subplot(1, 4, 2)
plt.title('Green_img')
plt.imshow(RGB_img)
RGB_img[:, :, 0] = Blue_img
RGB_img[:, :, 1] = Blue_img
RGB_img[:, :, 2] = Blue_img
plt.subplot(1, 4, 3)
plt.title('Blue_img')
plt.imshow(RGB_img)
plt.show()
```

```python
plt.figure(figsize=(10, 10))
plt.subplot(1, 3, 1)
plt.title('Red Img')
plt.imshow(Red_img, cmap='gray')
```
