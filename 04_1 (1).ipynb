```python
import cv2
import matplotlib.pyplot as plt

image_path = '/content/drive/My Drive/Colab/images/lena.jpg'

image = cv2.imread(image_path)

    r_channel = image[:, :, 2]

    plt.figure(figsize=(10, 5))
    plt.hist(r_channel.ravel(), bins=256, range=[0, 256], color='red', alpha=0.75)

    plt.title('Histogram of lena.jpg (Red Channel)')
    plt.xlabel('Pixel Intensity Value (0 - 255)')
    plt.ylabel('Frequency (Pixel Count)')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.xlim([0, 256])

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

image_path = '/content/drive/My Drive/Colab/images/cell.bmp'

gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    cv2_imshow(gray_image)

    plt.figure(figsize=(10, 5))
    plt.hist(gray_image.ravel(), bins=256, range=[0, 256], color='gray', alpha=0.75)

    plt.title('Histogram of cell.bmp (Grayscale)')
    plt.xlabel('Pixel Intensity Value (0 - 255)')
    plt.ylabel('Frequency (Pixel Count)')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.xlim([0, 256])

    plt.show()
```
