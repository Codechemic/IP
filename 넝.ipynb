```python
import cv2
import matplotlib.pyplot as plt

# 1. Google Drive 경로에서 이미지 로드 (MyDrive 대소문자 확인)
img = cv2.imread('/content/drive/MyDrive/Colab/images/images.webp')

# 2. 이미지 로드 성공 여부 확인
if img is not None:
    # 3. OpenCV의 BGR 색상 채널을 Matplotlib의 RGB 채널로 변환
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 4. plt.imshow()를 사용하여 이미지 표시
    plt.imshow(img_rgb)
    plt.axis('off')  # 이미지 테두리의 축(눈금) 제거
    plt.show()       # 화면에 최종 출력
else:
    print("이미지를 불러올 수 없습니다. 파일 경로 및 드라이브 마운트 상태를 확인해 주세요.")
```
