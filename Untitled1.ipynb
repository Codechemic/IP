```python
import cv2
import os
from google.colab.patches import cv2_imshow
from google.colab import drive

def custom_threshold(image_path, threshold_value):
    # 1. 이미지 읽기
    image = cv2.imread(image_path)
    if image is None:
        print(f"[오류] 이미지를 불러올 수 없습니다. 경로를 확인하세요: {image_path}")
        print("Google Drive가 마운트되지 않았거나 해당 경로에 파일이 존재하지 않을 수 있습니다.")
        return None

    # 2. 그레이스케일(흑백) 이미지로 변환
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # 3. 임계값 기준으로 이진화 수행 (임계값보다 크면 흰색(255), 아니면 검은색(0))
    _, thresholded = cv2.threshold(gray, threshold_value, 255, cv2.THRESH_BINARY)

    return image, thresholded

# 구글 드라이브 파일 경로
target_image_path = '/content/drive/My Drive/Colab/images/cell-bright.bmp'

# 드라이브 마운트 여부 확인 및 필요한 경우 마운트 시도
if not os.path.exists('/content/drive'):
    print("구글 드라이브 마운트를 진행합니다...")
    try:
        drive.mount('/content/drive')
    except Exception as e:
        print("드라이브 마운트 중 오류 발생:", e)

# 임계값을 127로 설정하여 이진화 실행
threshold_val = 127
results = custom_threshold(target_image_path, threshold_val)

if results is not None:
    original, binary_img = results
    print("원본 이미지:")
    cv2_imshow(original)
    print(f"이진화 결과 (기준 임계값: {threshold_val}):")
    cv2_imshow(binary_img)
```
