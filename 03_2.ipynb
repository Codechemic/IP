```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def apply_saturation_gamma(image, gamma=1.0):
    inv_gamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** inv_gamma) * 255 for i in np.arange(256)]).astype("uint8")
    return cv2.LUT(image, table)

gamma_value_high = 1.5
saturated_img1 = apply_saturation_gamma(img1, gamma=gamma_value_high)

gamma_value_low = 0.7
desaturated_img1 = apply_saturation_gamma(img1, gamma=gamma_value_low)

saturated_abs_diff_image = apply_saturation_gamma(abs_diff_image, gamma=gamma_value_high)

cv2_imshow(img1)

cv2_imshow(saturated_img1)

cv2_imshow(desaturated_img1)

cv2_imshow(abs_diff_image)

cv2_imshow(saturated_abs_diff_image)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

base_path = '/content/drive/My Drive/Colab/images/'
mask_file = 'circle.bmp'
image_to_mask_file = 'lena.jpg'

mask_path = base_path + mask_file
image_to_mask_path = base_path + image_to_mask_file

mask_image = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

original_image = cv2.imread(image_to_mask_path, cv2.IMREAD_COLOR)

    if mask_image.shape[:2] != original_image.shape[:2]:
        print("경고: 마스크와 원본 이미지의 크기가 다릅니다. 마스크 이미지에 맞춰 원본 이미지 크기를 조정합니다.")
        height, width = mask_image.shape[:2]
        original_image = cv2.resize(original_image, (width, height))
        print(f"이미지 크기 조정됨: {original_image.shape}")

    masked_image = cv2.bitwise_and(original_image, original_image, mask=mask_image)

    print(f"\n원본 마스크 이미지 ({mask_file}):")
    cv2_imshow(mask_image)

    print(f"\n원본 이미지 ({image_to_mask_file}):")
    cv2_imshow(original_image)

    print(f"\n마스크 적용 결과 (흰색 부분에 {image_to_mask_file} 이미지가 나옴):")
    cv2_imshow(masked_image)
```
